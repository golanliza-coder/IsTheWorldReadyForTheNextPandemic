## Re-arrainging the data in file "WHO-COVID-19-global-daily-data_md.csv":
1. Turning file "WHO-COVID-19-global-daily-data_md.csv" from date driven to year driven to alighn values
2. Print file EDA

```python
import os
import pandas as pd

# =====================================================================
# 1. הגדרת נתיב הספרייה ושם הקובץ (עדכנו את ה-PATH שלכן כאן)
# =====================================================================
FOLDER_PATH = '/content/drive/MyDrive/DS_IsTheWorldReadyForTheNextPandemic/DATASET TO USE/'  # הכניסו כאן את ה-PATH של הספרייה
FILE_NAME = 'WHO-COVID-19-global-daily-data_md.csv'         # שם הקובץ שלכן

# חיבור אוטומטי של הספרייה והקובץ לנתיב מלא אחד
FULL_PATH = os.path.join(FOLDER_PATH, FILE_NAME)

print(f"טוען את הקובץ מתוך הנתיב: {FULL_PATH}")

# =====================================================================
# 2. טעינת הקובץ וניקוי ראשוני
# =====================================================================
df = pd.read_csv(FULL_PATH)

# ניקוי רווחים מיותרים משמות העמודות ליתר ביטחון
df.columns = df.columns.str.strip()

# הפיכת עמודת התאריך לפורמט מתאים וחילוץ השנה לקבוצה
df['Date_reported'] = pd.to_datetime(df['Date_reported'], errors='coerce')
df['YEAR'] = df['Date_reported'].dt.year

# =====================================================================
# 3. הגדרת חוקי הקיבוץ (Aggregation Rules)
# =====================================================================
aggregation_rules = {
    'Country_code': 'first',      # שמירת קוד המדינה
    'WHO_region': 'first',        # שמירת האזור הגיאוגרפי
    'New_cases': 'sum',           # סכימת המקרים החדשים לאורך כל השנה
    'New_deaths': 'sum',          # סכימת מקרי המוות החדשים לאורך כל השנה
    'Cumulative_cases': 'max',    # הערך המקסימלי בשנה (סך הכל בסוף השנה)
    'Cumulative_deaths': 'max'    # הערך המקסימלי בשנה (סך הכל בסוף השנה)
}

# =====================================================================
# 4. ביצוע הקיבוץ ושמירת הקובץ השנתי באותה ספרייה
# =====================================================================
df_yearly = df.groupby(['Country', 'YEAR']).agg(aggregation_rules).reset_index()

# שמירת הקובץ החדש באותה הספרייה שהגדרתן למעלה
OUTPUT_NAME = 'yearly_covid_summary.csv'
OUTPUT_PATH = os.path.join(FOLDER_PATH, OUTPUT_NAME)
df_yearly.to_csv(OUTPUT_PATH, index=False)

print(f"הטבלה קובצה בהצלחה מרמה יומית לרמה שנתית!")
print(f"הקובץ המעובד נשמר בספרייה שלכן תחת השם: {OUTPUT_NAME}")
print(f"מימדי הטבלה החדשה: {df_yearly.shape}")
```

## Re-arrainging the data in file "epidemiology_Ind_md.csv":
1. Turning file "epidemiology_Ind_md.csv" from date driven to year driven to alighn values
2. Eliminating "Subregion" fields
3. Assign Null fields = 0
4. Print file EDA

```python
import os
import pandas as pd

# =====================================================================
# 1. הגדרת נתיב הספרייה ושם הקובץ האפידמיולוגי
# =====================================================================
FOLDER_PATH = '/content/drive/MyDrive/DS_IsTheWorldReadyForTheNextPandemic/DATASET TO USE/'  # עדכנו את ה-PATH שלכן כאן
FILE_NAME = 'epidemiology_Ind_md.csv'

FULL_PATH = os.path.join(FOLDER_PATH, FILE_NAME)
print(f"טוען את הקובץ מתוך: {FULL_PATH}")

# טעינת הקובץ וניקוי רווחים משמות העמודות
df = pd.read_csv(FULL_PATH)
df.columns = df.columns.str.strip()

# =====================================================================
# 2. הסרת עמודות D, E, F, G (לפי אינדקס המיקום שלהן בטבלה)
# =====================================================================
# עמודות D, E, F, G הן עמודות מספר 3, 4, 5, 6 (מתחילים לספור מ-0)
# אלו הן העמודות: subregion1_code, subregion1_name, subregion2_code, subregion2_name
columns_to_drop = [df.columns[3], df.columns[4], df.columns[5], df.columns[6]]
df = df.drop(columns=columns_to_drop)

# =====================================================================
# 3. טיפול בערכים חסרים: החלפת כל ה-NULL / NaN ב-0
# =====================================================================
df = df.fillna(0)

# חילוץ השנה מתוך עמודת התאריך לטובת הקיבוץ
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['YEAR'] = df['date'].dt.year

# =====================================================================
# 4. הגדרת חוקי הקיבוץ השנתי (Aggregation Rules)
# =====================================================================
# אגרגציה לפי לוגיקה של למידת מכונה: נתונים יומיים נסכום, נתונים מצטברים ניקח מקסימום
aggregation_rules = {
    'country_code': 'first',
    'country_name': 'first',
    'aggregation_level': 'first',
    'new_confirmed': 'sum',
    'new_deceased': 'sum',
    'new_recovered': 'sum',
    'new_tested': 'sum',
    'cumulative_confirmed': 'max',
    'cumulative_deceased': 'max',
    'cumulative_recovered': 'max',
    'cumulative_tested': 'max'
}

# =====================================================================
# 5. ביצוע הקיבוץ השנתי ושמירת הקובץ החדש
# =====================================================================
# מקבצים לפי מפתח המדינה (location_key) והשנה (YEAR)
df_yearly = df.groupby(['location_key', 'YEAR']).agg(aggregation_rules).reset_index()

# שמירת הקובץ המעובד באותה ספרייה
OUTPUT_NAME = 'yearly_epidemiology_cleaned.csv'
OUTPUT_PATH = os.path.join(FOLDER_PATH, OUTPUT_NAME)
df_yearly.to_csv(OUTPUT_PATH, index=False)

print("\nהפעולות בוצעו בהצלחה!")
print(f"1. עמודות D,E,F,G הוסרו.")
print(f"2. כל ערכי ה-NULL הוחלפו ב-0.")
print(f"3. הנתונים קובצו לרמה שנתית ונשמרו בשם: {OUTPUT_NAME}")
print(f"מימדי הטבלה השנתית החדשה: {df_yearly.shape}")
```

## Re-arrainging the data in file "vaccinations_Ind_md.csv":
Re-arrainging the data in file "vaccinations_Ind_md.csv":
1. Turning file "vaccinations_Ind_md.csv" from date driven to year driven to alighn values
2. Eliminating "Subregion" fields
3. Assign Null fields = 0
4. Print file EDA
##

```python
import os
import pandas as pd

# =====================================================================
# 1. הגדרת נתיב הספרייה ושם קובץ החיסונים
# =====================================================================
FOLDER_PATH = '/content/drive/MyDrive/DS_IsTheWorldReadyForTheNextPandemic/DATASET TO USE/'  # עדכנו את ה-PATH שלכן כאן
FILE_NAME = 'vaccinations_Ind_md.csv'

FULL_PATH = os.path.join(FOLDER_PATH, FILE_NAME)
print(f"טוען את קובץ החיסונים מתוך: {FULL_PATH}")

# טעינת הקובץ וניקוי רווחים משמות העמודות
df_vac = pd.read_csv(FULL_PATH)
df_vac.columns = df_vac.columns.str.strip()

# =====================================================================
# 2. הסרת עמודות D, E, F, G (לפי מיקום העמודה בטבלה)
# =====================================================================
# עמודות 3, 4, 5, 6 הן עמודות ה-subregion הריקות שרואים בתמונה
columns_to_drop = [df_vac.columns[3], df_vac.columns[4], df_vac.columns[5], df_vac.columns[6]]
df_vac = df_vac.drop(columns=columns_to_drop)

# =====================================================================
# 3. מילוי כל השדות הריקים (Null/NaN) ב-0
# =====================================================================
df_vac = df_vac.fillna(0)

# חילוץ השנה מתוך עמודת התאריך לטובת הקיבוץ השנתי
df_vac['date'] = pd.to_datetime(df_vac['date'], errors='coerce')
df_vac['YEAR'] = df_vac['date'].dt.year

# =====================================================================
# 4. בנייה אוטומטית של חוקי הקיבוץ (Aggregation Rules)
# =====================================================================
aggregation_rules = {
    'country_code': 'first',
    'country_name': 'first',
    'aggregation_level': 'first'
}

# סריקה אוטומטית של שאר העמודות בקובץ כדי להתאים sum או max
for col in df_vac.columns:
    if col in ['location_key', 'date', 'YEAR', 'country_code', 'country_name', 'aggregation_level']:
        continue

    # אם העמודה מייצגת נתון מצטבר, ניקח את הערך המקסימלי השנתי
    if 'cumulative' in col.lower() or 'total' in col.lower():
        aggregation_rules[col] = 'max'
    # אם העמודה מייצגת נתון יומי חדש, נסכום את כל השנה
    else:
        aggregation_rules[col] = 'sum'

# =====================================================================
# 5. ביצוע הקיבוץ (Groupby) ושמירת הקובץ השנתי המעובד
# =====================================================================
df_vac_yearly = df_vac.groupby(['location_key', 'YEAR']).agg(aggregation_rules).reset_index()

# שמירת הקובץ המעובד באותה ספרייה
OUTPUT_NAME = 'yearly_vaccinations_final_summary.csv'
OUTPUT_PATH = os.path.join(FOLDER_PATH, OUTPUT_NAME)
df_vac_yearly.to_csv(OUTPUT_PATH, index=False)

print("\n--- קובץ החיסונים עובד בהצלחה בהתאם לכל הדרישות! ---")
print(f"1. עמודות ה-subregion (D,E,F,G) הוסרו בהצלחה.")
print(f"2. כל השדות הריקים מולאו ב-0.")
print(f"3. עמודות מסוג Cumulative/Total קיבלו ערך מקסימלי, ועמודות מסוג New נסכמו שנתית.")
print(f"4. הקובץ השנתי נשמר בשם: {OUTPUT_NAME}")
print(f"מימדי הטבלה החדשה: {df_vac_yearly.shape}")
```

```python
# 1. מדדי מרכז ופיזור (ממוצע, חציון, סטיית תקן, מינימום, מקסימום ואחוזונים)
print("--- מדדי מרכז ופיזור עבור עמודות החיסונים והתגובה ---")
# שימוש ב-.T הופך את הטבלה כדי שיהיה קל לקרוא את המדדים עבור עמודות החיסונים הארוכות
display(df_vac_yearly.describe().T)

# 2. וידאו סופי שאין יותר ערכים חסרים (התוצאה בכל העמודות צריכה להיות 0)
print("\n--- בדיקת ערכים חסרים (Missing Values) בקובץ החיסונים המעובד ---")
missing_summary = pd.DataFrame({
    'כמות ערכי NULL': df_vac_yearly.isnull().sum(),
    'אחוז ערכי NULL': (df_vac_yearly.isnull().sum() / len(df_vac_yearly)) * 100
})
display(missing_summary)

# 3. מדדי התפלגות (צידוד וגבנוניות - Skewness & Kurtosis)
print("\n--- מדדי התפלגות מורחבים של עמודות החיסונים ---")
distribution_stats = pd.DataFrame({
    'Skewness (צידוד)': df_vac_yearly.select_dtypes(include=['number']).skew(),
    'Kurtosis (גבנוניות)': df_vac_yearly.select_dtypes(include=['number']).kurt()
})
display(distribution_stats)
```

## Merging File "epidemiology_Ind_md.csv" with file "demographic_Ind_md.csv" with file:

data.

3. Calculating X4 = New Tested / Population * 1000 (Normalized)
##

```python
import os
import pandas as pd

# (בהנחה ששני המשתנים df_yearly ו-df_demo_cleaned קיימים כרגע בזיכרון שלכן)

# =====================================================================
# 1. ניקוי עמודות כפולות מטבלת הדמוגרפיה לפני החיבור
# =====================================================================
cols_to_drop_from_demo = ['country_code', 'country_name', 'aggregation_level']
df_demo_filtered = df_demo_cleaned.drop(columns=[col for col in cols_to_drop_from_demo if col in df_demo_cleaned.columns])

# =====================================================================
# 2. ביצוע ה-Left Join לפי מפתח המדינה (location_key)
# =====================================================================
print("מבצע מיזוג שמאלי (Left Join) של נתוני אפידמיולוגיה ודמוגרפיה...")
df_epidemiology_demographics = pd.merge(
    df_yearly,          # טבלת הבסיס השנתית (האפידמיולוגית)
    df_demo_filtered,   # טבלת הדמוגרפיה המסוננת
    on='location_key',  # החיבור מתבצע לפי מפתח המדינה
    how='left'          # שמירה על כל שורות התחלואה והתאמת הדמוגרפיה אליהן
)

# =====================================================================
# 3. ניקוי שלאחר המיזוג: מילוי ערכים חסרים (NaN) מבוקר ולא גורף
# =====================================================================
# נמלא ב-0 רק עמודות של מונים אפידמיולוגיים יומיים/שנתיים שבהם NaN מייצג אפס מקרים
cols_to_fill_zero = ['new_confirmed', 'new_deceased', 'total_confirmed', 'total_deceased']
for col in cols_to_fill_zero:
    if col in df_epidemiology_demographics.columns:
        df_epidemiology_demographics[col] = df_epidemiology_demographics[col].fillna(0)

# שאר עמודות השיעורים, המדדים והתמותה העודפת נשארות כ-NaN כדי לא לעוות את הנתונים

# =====================================================================
# 4. שמירת הקובץ המאוחד החדש
# =====================================================================
FOLDER_PATH = '/content/drive/MyDrive/DS_IsTheWorldReadyForTheNextPandemic/DATASET TO USE/'
OUTPUT_NAME = 'merged_epidemiology_demographics_yearly.csv'
OUTPUT_PATH = os.path.join(FOLDER_PATH, OUTPUT_NAME)

df_epidemiology_demographics.to_csv(OUTPUT_PATH, index=False)

print(f"\n--- המיזוג (מיזוג 4 בתוכנית) הסתיים בהצלחה! ---")
print(f"הקובץ המאוחד נשמר בשם: {OUTPUT_NAME}")
print(f"מימדי הטבלה המאוחדת החדשה (שורות, עמודות): {df_epidemiology_demographics.shape}")
```

```python
import os
import pandas as pd

# הגדרת ה-PATH שלכן לשמירה ב-Google Drive
FOLDER_PATH = '/content/drive/MyDrive/DS_IsTheWorldReadyForTheNextPandemic/DATASET TO USE/'
OUTPUT_NAME = 'merged_epidemiology_demographics_yearly.csv'
OUTPUT_PATH = os.path.join(FOLDER_PATH, OUTPUT_NAME)

print("1. מכין את משתני הזיכרון המקוריים...")
# יצירת עותקים נקיים וניקוי רווחים נסתרים משמות העמודות
df_epi_base = df_yearly.copy()
df_epi_base.columns = df_epi_base.columns.str.strip()

df_demo_base = df_demo_cleaned.copy()
df_demo_base.columns = df_demo_base.columns.str.strip()

# =====================================================================
# 2. ניקוי עמודות כפולות מטבלת הדמוגרפיה לפני החיבור
# =====================================================================
cols_to_drop_from_demo = ['country_code', 'country_name', 'aggregation_level']
df_demo_filtered = df_demo_base.drop(columns=[col for col in cols_to_drop_from_demo if col in df_demo_base.columns])

# =====================================================================
# 3. ביצוע ה-Left Join לפי מפתח המדינה (location_key) - שמירה על כל העמודות
# =====================================================================
print("2. מבצע מיזוג שמאלי (Left Join) ושומר על כל העמודות המקוריות...")
df_epidemiology_demographics = pd.merge(
    df_epi_base,        # טבלת הבסיס השנתית האפידמיולוגית
    df_demo_filtered,   # טבלת הדמוגרפיה המסוננת
    on='location_key',  # החיבור מתבצע לפי מפתח המדינה
    how='left'          # שמירה על כל שורות התחלואה והתאמת הדמוגרפיה אליהן
)

# =====================================================================
# 4. חישוב משתנה X4 המדויק על בסיס העמודה new_tested
# =====================================================================
print("3. מחשב את משתנה X4 (סך בדיקות ל-1,000 איש) מתוך עמודת new_tested...")

# וידוא ששני השדות קיימים בטבלה המשולבת
if 'new_tested' in df_epidemiology_demographics.columns and 'population' in df_epidemiology_demographics.columns:
    # חישוב הנוסחה: (בדיקות חלקי אוכלוסייה) כפול 1,000
    df_epidemiology_demographics['X4'] = (df_epidemiology_demographics['new_tested'] / df_epidemiology_demographics['population']) * 1000

    # טיפול במצבים של חלוקה באפס (אם יש מדינה עם אוכלוסייה 0) או ערכים חסרים
    df_epidemiology_demographics['X4'] = df_epidemiology_demographics['X4'].fillna(0)
else:
    print("שגיאה: העמודות new_tested או population לא נמצאו בטבלה!")

# =====================================================================
# 5. ניקוי שלאחר המיזוג: מילוי ערכים חסרים (NaN) מבוקר (לא גורף)
# =====================================================================
# נמלא ב-0 רק עמודות של מונים אפידמיולוגיים יומיים/שנתיים
cols_to_fill_zero = ['new_confirmed', 'new_deceased', 'new_recovered', 'new_tested',
                     'cumulative_confirmed', 'cumulative_deceased', 'cumulative_recovered', 'cumulative_tested']

for col in cols_to_fill_zero:
    if col in df_epidemiology_demographics.columns:
        df_epidemiology_demographics[col] = df_epidemiology_demographics[col].fillna(0)

# עמודות המדדים, האחוזים והתמותה העודפת נשארות כפי שהן לטיפול בחציון בשלבים הבאים

# =====================================================================
# 6. שמירת הקובץ המתוקן והמלא ל-Google Drive
# =====================================================================
df_epidemiology_demographics.to_csv(OUTPUT_PATH, index=False)

print(f"\n--- הקובץ שוחזר, חושב ותוקן בהצלחה! ---")
print(f"הקובץ המלא נשמר בשם: {OUTPUT_NAME}")
print(f"מימדי הטבלה המשוחזרת (שורות, עמודות כולל X4): {df_epidemiology_demographics.shape}")
```
